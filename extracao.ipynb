# Análise de Sentimento
Este notebook aplica o módulo `sentiment_analyzer` na base de comentários do arquivo `NPS.xlsx`.

```python
import pandas as pd
from sentiment_analyzer import get_sentiment
```

```python
# Carregar a base de dados
df = pd.read_excel('NPS.xlsx', keep_default_na=False)
df.head(3)
```

```python
# Aplicar a análise de sentimento na coluna 'Comment'
# Como get_sentiment retorna uma tupla (polaridade, classificação), vamos desempacotar o resultado em duas colunas.
df[['Polarity', 'Sentiment']] = df['Comment'].apply(lambda x: pd.Series(get_sentiment(str(x))))
```

```python
# Visualizar as primeiras linhas atualizadas e o resumo dos sentimentos
display(df.head())
print('\nDistribuição dos Sentimentos:')
print(df['Sentiment'].value_counts())
```
